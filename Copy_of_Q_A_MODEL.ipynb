# QUESTION ANSWER MODEL THROUGH TRANSFORMERS

#### This project demonstrates a Question Answering (QA) system built using transformer-based models.
The system takes a context paragraph and a question as input, then extracts the most relevant answer.

It showcases the power of Natural Language Processing (NLP) and pre-trained transformer models.

## Load Pre-Trained Model

```python
from transformers import pipeline
qa_model=pipeline("question-answering",model="distilbert/distilbert-base-uncased-distilled-squad")
```

### DEFINE Q\A FUNCTION

```python
def answer_question(question,context):
  result=qa_model(question=question,context=context)
  print("❓ Question❓:",question)
  print("✅ Answer: ✅ ",result["answer"])
  print("confidence:",round(result['score'],2))

```

### CONTEXT(ARTIFICAL INTELLIGENCE)

```python
context="""Artificial Intelligence (AI) is a branch of computer science that focuses on creating machines capable of performing tasks that typically require human intelligence.
 These tasks include problem-solving, learning, reasoning, understanding language, and recognizing patterns. One of the most important areas of AI is Natural Language Processing (NLP), which enables machines to understand and generate human language.
 Transformer models, such as BERT and GPT, have significantly improved NLP tasks by using attention mechanisms to understand context in text.
 These models are widely used in applications like chatbots, sentiment analysis, text summarization, and question answering systems.
 AI is continuously evolving and is being applied in various industries including healthcare, education, finance, and entertainment."""
```

### 🤖 AI Model Prediction (Question Answering)

```python
answer_question("What are some tasks AI can perform?",context)
```

```python
answer_question("what is NLP?",context)
```

```python
answer_question("name two transformers model?",context)
```

```python
answer_question("Name some applications of AI?",context)
```
